## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target is whether a webpage is declining. I use trend_direction = "down" as the defined label. A value of 1 means declining and 0 means not declining.


```python
import pandas as pd
import numpy as np

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

df["is_declining_label"] = (
    df["trend_direction"].str.lower().eq("down").astype(int)
)

print(df.shape)
print("Declining rate:", round(df["is_declining_label"].mean(), 3))
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

I will use Precision@20 and Precision@50. A higher precision means more of the pages identified by the model are actually declining.


```python
def precision_at_k(scores, labels, k):
    order = np.argsort(-np.asarray(scores))
    top_k = np.asarray(labels)[order[:k]]
    return top_k.mean()

print("Metrics: Precision@20 and Precision@50")
```
